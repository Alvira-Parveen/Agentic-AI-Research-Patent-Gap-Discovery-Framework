```python
from src.embeddings.embedding_service import EmbeddingService
from src.retrieval.patent_retriever import PatentRetriever
from src.retrieval.paper_retriever import PaperRetriever

embedder = EmbeddingService.get_instance()
print(f"Active model: {embedder.model_name}, Dimension: {embedder.dimension}")

query = "drone multispectral disease classification"
p_ret = PatentRetriever()
r_ret = PaperRetriever()

p_res = p_ret.retrieve(query, top_k=3)
r_res = r_ret.retrieve(query, top_k=3)

print("--- TOP PATENTS ---")
for p in p_res:
    print(f"{p.citation}: {p.document_title} (Sim: {p.similarity_score:.4f})")

print("
--- TOP RESEARCH PAPERS ---")
for r in r_res:
    print(f"{r.citation}: {r.document_title} (Sim: {r.similarity_score:.4f})")
```
