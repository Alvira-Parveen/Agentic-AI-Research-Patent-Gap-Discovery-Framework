# PBL-3: Corpus Exploration and Data Preprocessing
This notebook demonstrates the inspection, text extraction, and section-aware chunking of the 20 USPTO patents and 20 research papers.

```python
import os
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from src.schemas.documents import DocumentType
from src.ingestion.metadata import load_processed_dataset

pat_meta, pat_chunks = load_processed_dataset(DocumentType.PATENT)
pap_meta, pap_chunks = load_processed_dataset(DocumentType.PAPER)

print(f"Patents Ingested: {len(pat_meta)} docs, {len(pat_chunks)} chunks")
print(f"Papers Ingested:  {len(pap_meta)} docs, {len(pap_chunks)} chunks")
```

```python
# Section Distribution
pat_sec_counts = pd.Series([c.section for c in pat_chunks]).value_counts()
pap_sec_counts = pd.Series([c.section for c in pap_chunks]).value_counts()

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
pat_sec_counts.plot(kind="bar", ax=axes[0], title="Patent Chunks by Section", color="steelblue")
pap_sec_counts.plot(kind="bar", ax=axes[1], title="Research Paper Chunks by Section", color="teal")
plt.tight_layout()
plt.show()
```
